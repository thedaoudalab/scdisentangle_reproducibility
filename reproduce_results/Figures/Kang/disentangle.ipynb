# Imports

```python
import os
import torch
import scanpy as sc
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from sklearn.feature_selection import mutual_info_regression
from scdisentangle.train.tools import get_trainer, set_seed
import warnings
```

```python
warnings.filterwarnings('ignore')
```

# Set seed

```python
seed_nb = 42
set_seed(seed_nb)
```

# Get Trainer

```python
yaml_path = 'configs/kang_disentangle.yaml'
```

```python
trainer = get_trainer(yaml_path, wandb_log=False, seed_nb=seed_nb)
```

# Train

```python
trainer.train()
```

# Load best weights

```python
weights_path = 'weights/MIG_BINNED_dis_latent_stack_condition_train'
trainer.load_weights(weights_path)
```

# Get Latent

```python
counterfactual_dict = {'condition': 'control'}
```

# Inference (intervene on condition to predict everything as control)

```python
adata = trainer.predict(
    trainer.dataset.data.copy(), 
    counterfactual_dict=counterfactual_dict, 
    bs=256
)
```

# Normalize

```python
sc.pp.normalize_total(adata)
sc.pp.log1p(adata)
```

# UMAP of predictions on expression space

```python
sc.pp.neighbors(adata)
sc.tl.umap(adata)
```

```python
sc.pl.umap(adata, color='cell_type')
```

```python
sc.pl.umap(adata, color='condition')
```

```python
sc.pl.umap(adata, color='condition_org')
```

# Save original versus predictions (batch corrected)

```python
adata_org = trainer.dataset.data.copy()
sc.pp.normalize_total(adata_org)
sc.pp.log1p(adata_org)
```

```python
sc.pp.neighbors(adata_org)
sc.tl.umap(adata_org)
```
